### #1.2

Crea una función similar a la que se pide en el ejercicio 6 de la segunda tanda de ejercicios del workout. Es decir, una función para pintar diagramas de dispersion que tenga seis argumentos, pero ojo al cambio en el quinto:
1. Un dataframe
2. Una columna numérica del dataframe
3. Una columna numérica del dataframe diferente a la anterior.
4. Un columna categórica (Opcional, valor por defecto None)
5. Un valor para el tamaño de los puntos o una cadena, pero que puede ser una cadena de caracteres, en cuyo caso se debe interpretar como una de las columnas del dataframe cuyos valores deben controlar el tamaño
6. Un sexto argumento "escala", que permita manipular la escala del tamaño en el caso de que el quinto argumento sea una cadena. Este argumento multiplicará a los valores de la columna indicada en el parámetro anterior (así si quiero dividir por 10000, este sexto argumento tendrá que valer 1/10000)

La función debe pintar el diagrama de dispersión de la columna numérica 1, en el eje X, frente a la columna numérica 2, en el eje Y, usando un código de colores en función de la columna categórica y con el tamaño de los puntos según el argumento correspondiente. 

Si no se diera una columna categórica como argumento, el color no debe pasarse como parámetro a la función de Seaborn o Matplotlib que uses.

EXTRA OPCIONAL: Añade un argumento que controle si la función pinta una leyenda para colores y tamaños o un colorbar y una leyenda para los tamaños.

```python
def diagrama_dispersion(df, var1, var2, var_cat=None, size=None, scale=1):
    plt.figure(figsize=(20, 10))

    if isinstance(size, str):    # Si queremos tamaños según los valores de una columna
        tam_puntos = df[size] * scale
    elif size is not None:  # Tamaño fijo dado
        tam_puntos = size

    if var_cat is None:
        sns.scatterplot(data=df, x=var1, y=var2, size=tam_puntos, alpha=0.5)    # Sin categórica
    else:
        sns.scatterplot(data=df, x=var1, y=var2, hue=var_cat, size=tam_puntos, palette="viridis", alpha=0.5)    # Con categórica

    plt.title(f" Diagrama de dispersión de {var1} frente a {var2}")
    plt.xlabel(var1)
    plt.ylabel(var2)
    plt.show()

    return

```

Y generar el notebook con las gráficas que vas a emplear en la presentación. Tienes que entregar ambas cosas.

```python
df_titanic = pd.read_csv("./data/titanic.csv")
```

```python
# Mortalidad
tasas = df_titanic["alive"].value_counts(normalize=True) * 100
plt.figure(figsize=(6, 4))

sns.barplot(x=tasas.index, y=tasas.values, hue=sorted(df_titanic.alive.unique()),  palette="Set1")

plt.title("Supervivencia en el Titanic")
plt.ylabel("Porcentaje (%)")
plt.show()
```

```python
# Supervivencia por grupos
tasas = df_titanic.groupby(["who"])["survived"].mean() * 100
plt.figure(figsize=(7, 4))

sns.barplot(x=tasas.index, y=tasas.values, hue=sorted(df_titanic.who.unique()),  palette="icefire")

plt.title("Supervivencia por grupo")
plt.ylabel("Supervivencia (%)")
plt.show()
```

```python
# Supervivencia por clase
tasas = df_titanic.groupby("class", observed=False)["survived"].mean() * 100
plt.figure(figsize=(7, 4))

sns.barplot(x=tasas.index, y=tasas.values, hue=sorted(df_titanic["class"].unique()), palette="viridis")

plt.title("Supervivencia por clase")
plt.xlabel("Clase")
plt.ylabel("Supervivencia (%)")
plt.show()
```

```python
# Supervivencia clase grupo
plt.figure(figsize=(8, 5))

sns.barplot(data=df_titanic, x="class", y="survived", hue="who", palette="icefire", errorbar=None)

plt.title("Supervivencia por clase y grupo")
plt.xlabel("Class")
plt.ylabel("Tasa de supervivencia")
plt.show()
```

```python
# Supervivencia por embarque pero es más la clase
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.barplot(data=df_titanic, x="embark_town", y="survived", hue="embark_town", palette="viridis", errorbar=None, ax=axes[0])

axes[0].set_title("Supervivencia por puerto de embarque")
axes[0].set_xlabel("")
axes[0].set_ylabel("Tasa de supervivencia")


sns.barplot(data=df_titanic, x="embark_town", y="survived", hue="class", palette="rocket", errorbar=None, ax=axes[1])

axes[1].set_title("Clase del pasajero por puerto de embarque y clase")
axes[1].set_xlabel("")
axes[1].set_ylabel("Tasa de supervivencia")

plt.tight_layout()
plt.show()
```
